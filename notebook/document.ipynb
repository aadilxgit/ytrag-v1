```python
sample_texts = {
    "../data/text_files/python_intro.txt": """\
Python Programming Introduction
==============================

Python is a general-purpose programming language designed to be readable.
People use it to automate tasks, build websites, analyze data, and create
machine-learning applications. A Python program is a set of instructions that
the Python interpreter runs from top to bottom.

Values and variables
--------------------
A variable is a name that refers to a value. Python does not require you to
declare a variable's type in advance.

name = "Ada"
age = 12
is_learning = True

Common built-in types include str for text, int for whole numbers, float for
decimal numbers, and bool for True or False values. Use print(value) to display
a value. Use type(value) to inspect its type.

Making decisions and repeating work
------------------------------------
An if statement chooses what to do based on a condition. An else branch handles
the other case. Indentation marks which statements belong to each block.

temperature = 18
if temperature < 20:
    print("Bring a jacket")
else:
    print("A light shirt is fine")

A for loop repeats an action for each item in a sequence:

for language in ["Python", "JavaScript", "Rust"]:
    print(language)

Functions
---------
A function gives a reusable name to a group of instructions. Parameters provide
its inputs, and return sends a result back to the caller.

def greet(person):
    return f"Hello, {person}!"

message = greet("Ada")
print(message)

Collections
-----------
A list stores items in order and can be changed. A dictionary stores values
under named keys.

colors = ["blue", "green", "red"]
student = {"name": "Ada", "grade": 7}
print(colors[0])
print(student["name"])

Errors and modules
------------------
When something goes wrong, Python reports an exception. Read the error message
to find the problem's type and the line where it occurred. A try/except block
can handle an expected error without stopping the whole program.

Modules are reusable Python files. Import one to use its functions; for example,
import math lets you call math.sqrt(25). Python also includes a standard library
of modules for common tasks such as working with dates, files, and JSON.

Good first habits
-----------------
Use descriptive variable and function names, keep indentation consistent, and
test a small piece of code before adding more. Start by changing the examples:
try a different name, add another color, or write a function that adds two
numbers. Programming becomes easier when you run code, observe the result, and
make one change at a time.
""",
    "../data/text_files/machine_learning.txt": """\
Introduction to Machine Learning
================================

Machine learning (ML) is a way to build computer programs that learn patterns
from examples. Instead of writing every rule by hand, a developer gives an
algorithm data and a goal. The trained model can then make predictions about
new data. Machine learning is one area of artificial intelligence.

Data, features, and labels
-------------------------
A dataset is a collection of examples. Each example has features: the
information used to make a prediction. In a house-price dataset, features
might include the number of bedrooms and the house's location. A label is the
answer the model should learn to predict, such as the house's sale price.

Common types of machine learning
--------------------------------
In supervised learning, examples include the correct answers. A model can
learn to sort emails into spam and not-spam, or predict a price.

In unsupervised learning, examples have no supplied answers. A model looks for
groups or patterns, such as customers with similar shopping habits.

In reinforcement learning, a system takes actions and receives rewards or
penalties. It learns which actions work well over time.

The basic workflow
------------------
First, collect relevant data and check that it is accurate. Next, split the
data into training examples and test examples. The model learns from the
training data. The test data checks how well it handles examples it has not
seen before. After evaluation, the model can be used to make predictions.

Training and prediction
-----------------------
Training is the process of finding patterns in example data. Prediction, also
called inference, is using the trained model on new data.

In Python, a machine-learning library often provides a model with fit and
predict methods. fit learns from training examples; predict produces answers
for new examples. The exact code depends on the library and the problem.

Avoiding common mistakes
------------------------
A model can memorize its training examples instead of learning a useful
general pattern. This is called overfitting. Testing on separate data helps
reveal it. Also check for missing values, incorrect labels, and data that does
not represent the real problem. A more complicated model is not always better.

Machine learning is useful when examples contain patterns that are difficult
to express as simple rules. It still depends on people to choose the goal,
prepare trustworthy data, check results, and decide whether predictions are
appropriate to use.
"""
}

for filepath, content in sample_texts.items():
    with open(filepath, 'w', encoding="utf-8") as f:
        f.write(content)


print("Sample Text Files Created!")
```

```python
from langchain_community.document_loaders import TextLoader

loader= TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
document= loader.load()
print(document)
```

```python
### directory Loader for text files

from langchain_community.document_loaders import DirectoryLoader

#load all text files from a directory 

dir_loader= DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt", # pattern to match file names
    loader_cls= TextLoader,
    loader_kwargs= {'encoding': 'utf-8'},
    show_progress= True 
)

documents= dir_loader.load()
documents
```

```python
### directory Loader for pdf files

from langchain_community.document_loaders import PyMuPDFLoader

#load all text files from a directory 

dir_loader= DirectoryLoader(
    "../data/pdf",
    glob="**/*.pdf", # pattern to match file names
    loader_cls= PyMuPDFLoader,
    show_progress= True 
)

pdf_documents= dir_loader.load()
pdf_documents
```
